```python
# selecting and filtering data

import pandas as pd

data = {
    "student": ["Amina", "Brian", "Carol", "David"],
    "assignment": [18, 15, 20, 12],
    "cat": [20, 17, 24, 14],
    "exam": [65, 52, 80, 45]
}

df = pd.DataFrame(data)

# selecting a single column

print(df["exam"])
type(df["exam"])

# selecting multiple columns

print(df[["exam", "cat"]])

# selecting rows with .iloc

#integer based location
#to get the first row

print(df.iloc[0])

# selecting multiple rows with iloc

print(df.iloc[0:2])

# .iloc can select rows and columns
# the syntax is
#  df.iloc[row_position, column_position]

print(df.iloc[0,3])

# select using .loc
# label-based selection

print(df.loc[0])
print(df.loc[0, "exam"])




# filtering

print(df["exam"] > 60)
print(df[df["exam"] > 60])

# filtering with multiple conditions

print(df[(df["exam"] > 60) & (df["cat"] > 18)])

print(df[(df["exam"] > 70) | (df["exam"] < 50)])

# filtering and selecting column

print(df[df["exam"] > 60]["student"])

# you can also use .loc

print(df.loc[df["exam"] > 60, "student"])
```

```python
##  creating new columns and Feature engineering

import pandas as pd

import numpy as np

data = {
    "student": ["Amina", "Brian", "Carol", "David"],
    "assignment": [18, 15, 20, 12],
    "cat": [20, 17, 24, 14],
    "exam": [65, 52, 80, 45]
}

df = pd.DataFrame(data)

# creating a new column

df["total"] = df["assignment"] + df["cat"] + df["exam"]

df["exam_decimal"] = df["exam"] / 100

# creating boolean column

df["passed"] = df["total"] >= 100

# creating categorical information

df["status"] = np.where(
    df["total"] >= 100,
    "Excellent",
    "Below Excellent"
)

print(df)

```

```python
# aggregation and groupby()

import pandas as pd

data = {
    "student": ["Amina", "Brian", "Carol", "David"],
    "course": ["AI", "Web", "AI", "Web"],
    "assignment": [18, 15, 20, 12],
    "cat": [20, 17, 24, 14],
    "exam": [65, 52, 80, 45]
}

df = pd.DataFrame(data)

# sum
print(df["exam"].sum())
# average
print(df["exam"].mean())
#smallest value
print(df["exam"].min())
#largest value
print(df["exam"].max())
#count
print(df["exam"].count())
# median
print(df["exam"].median())

# std
print(df["exam"].std())

# aggregating an entire dataframe

print(df.mean(numeric_only=True))

#groupby

print(df.groupby("course")["exam"].mean())

# multiple aggregations at once

print(df.groupby("course")["exam"].agg(["mean", "max", "min", "count"]))

#counting categories

print(df["course"].value_counts())
```

```python
# missing data

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "student": ["Amina", "Brian", "Carol", "David", "Eve"],
    "assignment": [18, 15, np.nan, 12, 20],
    "cat": [20, np.nan, 24, 14, 22],
    "exam": [65, 52, 80, np.nan, 75]
})

print(df)

print(df.isnull())

# find missing value in one column

print(df["exam"].isnull())
print(df[df["exam"].isnull()])

# count missing values
print(df.isnull().sum())

# remove missing data
# removes row if one column is missing
clean_df = df.dropna()
print(clean_df)

# removes a row if all columns are missing
clean_df2 = df.dropna(how="all")

# removing row based n specific column

clean_df3 = df.dropna(subset=["exam"])


# filling missing values

df["cat"] = df["cat"].fillna(0)

# filling with mean

mean_assignment = df["assignment"].mean()

df["assignment"] = df["assignment"].fillna(mean_assignment)

# filling categorical values

df["department"] = df["department"].fillna("unknown")

```

```python
# data cleaning

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "student": [" Amina ", "Brian", "Carol", "David", "Amina "],
    "course": ["AI", "ai", "Web", "WEB", "AI"],
    "age": ["20", "21", "twenty", "22", "20"],
    "exam": [65, 52, 80, 150, 65],
    "email": [
        "amina@gmail.com",
        "brian@gmail.com",
        "carol@gmail.com",
        "david@gmail.com",
        "amina@gmail.com"
    ]
})

print(df)

# problem 1: Extra spaces
# problem 2: Inconsistent categories
# problem 3: Incorrect data type
# problem 4: Impossible value
# problem 5: Duplicate record

# step 1: inspect the data

print(df.info())
print(df.describe())


# clean extra spaces

df["student"] = df["student"].str.strip()
# df["student"] = df["student"].str.lower()

# standardizing categories
print(df["course"].value_counts())

df["course"] = df["course"].str.upper()

print(df["course"].value_counts())


# understanding data types

print(df.dtypes)

# converting text to numbers

df["age"] = pd.to_numeric(df["age"], errors="coerce")
# introduces null values

# finding invalid values

print(df[(df["exam"] < 0) | (df["exam"] > 100)])

# you can convert it to missing

df.loc[df["exam"] > 100, "exam"] = np.nan

# removing duplicate rows

print(df.duplicated())
print(df[df.duplicated()])

df = df.drop_duplicates()

# duplicates based on specific columns

print(df.duplicates(subset=["email"]))

df = df.drop_duplicates(subset=["email"])
```

```python
# sorting , ranking and selecting records

import pandas as pd

df = pd.DataFrame({
    "student": ["Amina", "Brian", "Carol", "David", "Eve", "Frank"],
    "course": ["AI", "Web", "AI", "Web", "AI", "Web"],
    "exam": [65, 52, 80, 45, 75, 90],
    "cat": [20, 17, 24, 14, 22, 26]
})

df["total"] = df["exam"] + df["cat"]

print(df)

print(df.sort_values("exam"))
print(df.sort_values("exam", ascending = False))

# sort multiple columns

print(df.sort_values(
    by=["course", "exam"],
    ascending = [True, False]
))

# select the top records using nlargest

print(df.nlargest(3, "exam"))

#or 
print(df.sort_values("exam", ascending=False).head(3))

# bottom records

print(df.nsmallest(2, "exam"))

# find the maximum and minimum values

print(df["exam"].max())

# Returns the highest exam mark, which is 90.But this returns only the value—not the student who achieved it.To retrieve the complete row

print(df.loc(df["exam"].idxmax()))
print(df.loc[df["exam"].idxmin()])

# rank students 

df["exam_rank"] = df["exam"].rank(ascending = False)

# handling ties 
df["rank"] = df["exam"].rank(
    ascending=False,
    method="dense"
)
```
