```python
import pandas as pd

data = {
    "student": ["Amina", "Brian", "Carol"],
    "assignment": [18, 15, 20],
    "cat": [20, 17, 24],
    "exam": [65, 52, 80]
}

df = pd.DataFrame(data)

print(df)

# A DataFrame is a two-dimensional table with rows and columns.
#Think of it like a database table or an Excel worksheet.
#A Series is a one-dimensional labeled array. A DataFrame is made up of columns that are Series

# inspect a dataset

print(df.head())
print(df.head(2))
print(df.tail(2))
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```
