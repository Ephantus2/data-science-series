```python
# apply, map, replace
import pandas as pd

df = pd.DataFrame({
    "student": ["Amina", "Brian", "Carol", "David", "Eve"],
    "gender": ["F", "M", "F", "M", "F"],
    "score": [78, 65, 91, 48, 83]
})
# map
df["gender"] = df["gender"].map({
    "F": "Female",
    "M": "Male"
})

# replace
df["gender"] = df["gender"].replace({
    "F": "Female",
    "M": "Male"
})

# apply

def classify(score):
    if score >= 70:
        return "Pass"
    else:
        return "Fail"

df["result"] = df["score"].apply(classify)

# using lambda functions

df["result"] = df["score"].apply(
    lambda score: "Pass" if score >= 70 else "Fail"
)
```

```python
 # pd.cut() and pd.qcut()

import pandas as pd

df = pd.DataFrame({
    "student": ["Amina", "Brian", "Carol", "David", "Eve", "Frank"],
    "score": [35, 48, 56, 72, 85, 96]
})

print(df)

# pd.cut
# define the bins

bins = [0, 50, 70, 85, 100]

# define the labels

labels = [
    "Fail",
    "Average",
    "Good",
    "Excellent"
]

df["performance"] = pd.cut(
    df["score"],
    bins = bins,
    labels = labels,
    include_lowest = True,
    right = False
)

print(df)
print(df["performance"].value_counts())
print(df["performance"].value_counts(
    normalize=True,
    sort=False
) * 100)

# pd.qcut

df = pd.DataFrame({
    "student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "score": [10, 20, 30, 40, 50, 60, 70, 80]
})

df["score_group"] = pd.qcut(
    df["score"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(df)
```

```python
# Building a complete data preparation pipeline

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "study_hours": [2, 4, 5, 7, 8, 10, 3, 6, 9, 4],
    "attendance": [50, 60, 70, 80, 90, 100, 55, 75, 95, 65],
    "previous_marks": [35, 45, 50, 65, 75, 90, 40, 60, 85, 48],
    "assignment_marks": [38, 47, np.nan, 66, 76, 91, 42, 62, 88, 50],
    "department": [
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS"
    ],
    "final_marks": [38, 48, 53, 67, 78, 92, 43, 63, 89, 51]
})

print(df)

# setp 1: separate features and targets

x = df.drop(columns=["final_marks"])

y = df["final_marks"]

print(x.head())
print(y.head())

# step 2: split the data before learning preprocessing values

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

# handle missing values with simpleImputer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

imputer = SimpleImputer(strategy="mean")


imputer.fit(x_train[["assignment_marks"]])

x_train[["assignment_marks"]] = imputer.transform(
    x_train[["assignment_marks"]]
)

x_test[["assignment_marks"]] = imputer.transform(
    x_test[["assignment_marks"]]
)

# encode categorical features

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(x_train[["department"]])

train_department = encoder.transform(
    x_train[["department"]]
)

test_department = encoder.transform(
    x_test[["dapartment"]]
)

# scale numerical features

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(x_train[["study_hours", "attendance"]])

x_train[["study_hours", "attendance"]] = scaler.transform(
    x_train[["study_hours", "attendance"]]
)

x_test[["study_hours", "attendance"]] = scaler.transform(
    x_test[["study_hours", "attendance"]]
)

# we prepared everything manually

'''This is where Scikit-learn pipelines become useful.

A pipeline allows you to combine preprocessing and model training into a single workflow.

Let's build one. '''

# build a complete preprocessing pipeline

# step 1: identify the numerical and categorical columns

numeric_features = [
    "study_hours",
    "attendance",
    "previous_marks",
    "assignment_marks"
]

categorical_features = [
    "department"
]

# step 2: create the numerical pipeline

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

# step 3: create the categorical pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent"))
    ("encoder", OneHotEncoder(handle_unknown = "ignore"))
])

# step 4: combine both pipelines

from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# step 5: create the complete machine learning pipeline

from sklearn.linear_model import LinearRegression

model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

# Train the complete pipeline

x = df.drop(columns=["final_marks"])
y = df["final_marks"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

model_pipeline.fit(x_train, y_train)


# step 7: Evaluate the model

from sklearn.metrics import mean_absolute_error, mean_squared_error

predictions = model_pipeline.predict(x_test)

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)

print("MAE:", mae)
print("MSE:", mse)

# part 8: preventing Data leakage

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

model_pipeline.fit(x_train, y_train)

predictions = model_pipeline.predict(x_test)

```
