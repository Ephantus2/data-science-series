```python
# part 1: measure of central tendancy
import pandas as pd
import numpy as np

# mean

marks = [45, 50, 55, 60, 65, 70, 75, 80, 85, 95]
#using numpy
mean = np.mean(marks)

print(mean)

#using pandas

df = pd.DataFrame({
    "marks": [40, 50, 60, 70, 80]
})

print(df["marks"].mean())

# median

marks = np.array([20, 30, 40, 50, 60, 70])
# using numpy
print(np.median(marks))

# using pandas

print(df["marks"].median())

# mode

marks = pd.Series([50, 60, 60, 70, 80])

print(marks.mode())


```

```python
# part 2: measure of spread

# range
# range = maximum - minimum

import numpy as np

group_a = np.array([48, 49, 50, 51, 52])
group_b = np.array([10, 30, 50, 70, 90])

range_a = np.max(group_a) - np.min(group_a)
range_b = np.max(group_b) - np.max(group_b)

print("Group A range:", range_a)
print("Group B range:", range_b)

# variance
# measuring deviations from the mean
""" 
Calculate the mean.
Calculate how far each observation is from the mean.
Square each difference.
Average the squared differences, using the appropriate denominator.
"""

# calculating variance with numpy

data = np.array([2,4,6])

population_variance = np.var(data)
sample_variance = np.var(data, ddof=1)

print("Population variance:", population_variance)
print("Sample variance:", sample_variance)

# standard deviation

data = np.array([2, 4, 6])

print("Population std: ", np.std(data))
print("Sample std: ", np.std(data, ddof=1))

```

```python


#part 3: percentiles and quartiles

# calculating percentiles in Numpy

import numpy as np
import pandas as pd

marks = np.array([40, 50, 55, 60, 65, 70, 75, 80, 85, 95])

p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

print("25th percentile:", p25)
print("50th percentile:", p50)
print("75th percentile:", p75)

#using pandas

marks = pd.Series([40, 50, 55, 60, 65, 70, 75, 80, 85, 95])
print(marks.quantile(0.25))
print(marks.quantile(0.50))
print(marks.quantile(0.75))

# quartiles

#visuaizing distributions

import matplotlib.pyplot as plt
import numpy as np

marks = np.array([
    45, 50, 52, 55, 57, 58, 60, 61, 62, 63,
    65, 66, 68, 70, 72, 74, 75, 78, 80, 85
])

plt.figure(figsize=(8,5))
plt.hist(marks, bins=6, edgecolor="black")

plt.xlabel("Examination marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Examination Marks")

plt.show()

```
