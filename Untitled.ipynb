```python
import numpy as np

# Boolean Filtering

scores = np.array([45, 67, 82, 39, 91, 74])

print(scores[scores >= 70])

# Numpy aggregation

scores = np.array([70, 80, 90, 60])

# sum

print(np.sum(scores))

# mean

print(np.mean(scores))

# minimum

print(np.min(scores))

# maximum

print(np.max(scores))

# standard deviation

print(np.std(scores))
```

```python
scores = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [90, 95, 100]
])

# average of each row

print(np.mean(scores, axis=1))

# average of each column

print(np.mean(scores, axis=0))

```

```python
# Broadcasting

import numpy as np

scores = np.array([70, 80, 90])

print(scores + 5)

# Broadcasting with 2 arrays

scores = np.array([70, 80, 90])
bonus = np.array([2,3,4])

print(scores + bonus)

# Broadcasting with 2D arrays

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(matrix + 5)

# reshaping arrays

values = np.array([1,2,3,4,5,6])
print(values.reshape(2,3))
print(values.reshape(3,2))
```
