```python
import numpy as np

# Boolean Filtering

scores = np.array([45, 67, 82, 39, 91, 74])

print(scores[scores >= 70])

# Numpy aggregation

scores = np.array([70, 80, 90, 60])

# sum

print(np.sum(scores))

# mean

print(np.mean(scores))

# minimum

print(np.min(scores))

# maximum

print(np.max(scores))

# standard deviation

print(np.std(scores))
```

```python
scores = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [90, 95, 100]
])

# average of each row

print(np.mean(scores, axis=1))

# average of each column

print(np.mean(scores, axis=0))

```
