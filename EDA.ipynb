```python
# Exploratory Data analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "study_hours": [
        2, 4, 5, 7, 8, 10, 3, 6, 9, 4,
        5, 7, 8, 2, 6, 9, 3, 10, 5, 7
    ],

    "attendance": [
        50, 60, 70, 80, 90, 100, 55, 75, 95, 65,
        72, 85, 92, 45, 78, 96, 58, 98, 68, 82
    ],

    "previous_marks": [
        35, 45, 50, 65, 75, 90, 40, 60, 85, 48,
        55, 70, 78, 30, 62, 88, 42, 95, 52, 68
    ],

    "assignment_marks": [
        38, 47, 52, 66, 76, 91, 42, 62, 88, 50,
        57, 72, 80, 33, 64, 90, 44, 96, 54, 69
    ],

    "department": [
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS",
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS"
    ],

    "final_marks": [
        38, 48, 53, 67, 78, 92, 43, 63, 89, 51,
        57, 72, 81, 34, 65, 91, 45, 97, 55, 70
    ]
})

# understand the structure of the dataset

# step 1: Examine the first rows

print(df.head())

# step 2: check the dataset dimensions

print(df.shape)

# step 3: Examine datatypes and missing values

print(df.info())

#  Generate statistical summaries

print(df.describe())

# you can also examine categorical features

print(df.describe(include="str"))

# check missing values

print(df.isna().sum())

# univariate analysis - study one variable at a time

# create a histogram of final marks

plt.figure(figsize=(8,5))

plt.hist(df["final_marks"], bins=8, edgecolor="black")

plt.title("Distribution of Final Marks")
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")

plt.show()

# Box plot: Investigate spread and potential outliers

plt.figure(figsize=(8,4))

sns.boxplot(x=df["final_marks"])

plt.title("Box plot of Final Marks")
plt.xlabel("Final Marks")

plt.show()

# bivariate analysis - investigate relationships between two variables

# scatter plot: study hours versus Final marks

plt.figure(figsize=(8,5))

plt.scatter(
    df["study_hours"],
    df["final_marks"]
)

plt.title("study Hours vs Final Marks")
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")

plt.show()

# investigate other relationships

plt.figure(figsize=(8, 5))

plt.scatter(
    df["attendance"],
    df["final_marks"]
)

plt.title("Attendance vs Final Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")

plt.show()

# add previous marks

plt.figure(figsize=(8, 5))

plt.scatter(
    df["previous_marks"],
    df["final_marks"]
)

plt.title("previous marks vs final marks")
plt.xlabel("previous marks")
plt.ylabel("final marks")

plt.show()

# correlation analysis

numerical_df = df.select_dtypes(include="number")

correlation_matrix = numerical_df.corr()

print(correlation_matrix.round(2))

# you can visualize it using a heatmap

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("correlation Matrix")

plt.show()

# Multivariate analysis - investigate several variables together

# pair plot

sns.pairplot(
    df[
        [
            "study_hours",
            "attendance",
            "previous_marks",
            "final_marks"
        ]
    ]
)

plt.show()


# Example: compare departments
# suppose you want to investigate wheather final marks differ between departments

# a box plot is suitable

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="department",
    y="final_marks"
)

plt.title("Final Marks by Department")
plt.xlabel("Department")
plt.ylabel("Final Marks")

plt.show()
```
